```python
import numpy as np
import matplotlib.pyplot as plt

# Beta values
beta = np.linspace(0.1, 10, 100)

# Steady-state responses
v_open = 1 / beta
v_P = 10 / (10 + beta)
v_PI = np.ones_like(beta)

# Plot
plt.figure()
plt.plot(beta, v_open, label="Open Loop")
plt.plot(beta, v_P, label="P Control")
plt.plot(beta, v_PI, label="PI Control")

plt.xlabel(r"$\beta$")
plt.ylabel(r"$v_{ss}/v_{\mathrm{ref}}$")
plt.title("Steady-State Velocity Response vs. Model Uncertainty")
plt.legend()
plt.grid()

# Save figure
save_path = r"C:\Users\zsong\Desktop\LaTex\Fall 2026 Notes\Feedback Controls\HW\HW1\HW1_4c.png"
plt.savefig(save_path, dpi=300, bbox_inches="tight")

plt.show()
```
