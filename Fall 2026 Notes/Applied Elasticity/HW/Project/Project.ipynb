### Integrating for Neutral Axis

```python
import numpy as np
from scipy.integrate import quad

# Geometry (mm)
rbar = 962.1
do = 114.3
t = 7.62

ri = rbar - do/2
ro = rbar + do/2
di = do - 2*t

A = np.pi/4 * (do**2 - di**2)

# Integrands
def f_outer(r):
    return 2*np.sqrt(max(0, (do/2)**2 - (r-rbar)**2))/r

def f_inner(r):
    return 2*np.sqrt(max(0, (di/2)**2 - (r-rbar)**2))/r

# Numerical integration
I_outer = quad(f_outer, ri, ro)[0]
I_inner = quad(f_inner, ri+t, ro-t)[0]

Rn = A / (I_outer - I_inner)
e = rbar - Rn

print(f"Neutral axis radius: {Rn:.4f} mm")
print(f"Eccentricity: {e:.4f} mm")
```
